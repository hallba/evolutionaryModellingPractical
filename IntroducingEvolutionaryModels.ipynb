## Motivation- evolutionary competition in the aging body

![image.png](Images/Schematic.png)

*Structure of the stratified squamous epithelia from the interfollicular epidermis (skin) and esophagus of adult mice. Proliferating keratinocytes are located in the basal layer. Upon differentiation they migrate through suprabasal layers until they are ultimately lost by shedding. A balance should be established between cell division and cell loss to guarantee tissue homeostasis. HF, hair follicle; SG, sebaceous gland. Taken from Piedrafita et al, 2020)*

Many of the tissues in our adult body are supported by populations of competing stem cells. Cells divide and are lost whilst maintaining the total size of the stem cell population at a constant level. This constant process of proliferation and differentation in tissues like the skin and oesophagus, known as *squamous epithelial tissues* is necessary to maintain the non-dividing, protective suprabasal layers that make up the majority of the tissue. Despite survival being a competitive process, unmutated cells are equally fit and have the same chance of persisting.

As we age however, our cells can gain mutations that give them a selective advantage. This enables clonal expansion and survival. When healthy tissues in older people have been sequenced they have been found to be full of mutations associated with cancer, but that have been tolerated in the healthy tissue. We study the competition between cells carrying different mutations to better understand this precancerous state and to comprehend the initation of cancer.

Piedrafita et al. A single-progenitor model as the unifying paradigm of epidermal and esophageal epithelial maintenance in mice. Nat Commun 11, 1429 (2020). https://doi.org/10.1038/s41467-020-15258-0

## Genetic drift

Consider an experiment where you have a population of dividing cells in a tissue. In adult bodies, these populations mostly exist at homeostasis, where the total number of cells remain constant. In a tissue this could be a stem cell niche, such as basal layer of squamous epthelial tissues like the oesophagus or skin. In the experiment you introduce a label to 1 % of cells that once acquired is inherited by all descendents of the cell- that is, all of the cells daughters, and all of their daughters. Other than labelling the cell, the experiment does not change any other properties of the cell. You collect several replicates.

You return several divisions later to see how the populations of labelled cells have changed. Despite the average number of cells per replicate remaining homeostatic, each replicate has different sized labelled populations. This arises because though no individual cells have an advantage, the cells are in competition. This creates random opportunities for individual cells to expand, whilst others are replaced and do not survive in the tissue. This creates variation between different replicates known as *genetic drift*. This concept sits alongside the *Hardy-Weinberg equilibrium* that states that in the absence of evolutionary pressure the allele frequency remains unchanged, but introduces the impact of *stochastic* variation that arises in finite populations and can lead to substantial changes in properties of labelled populations.

### Case study of genetic drift- the single progenitor model of epithelial squamous tissues

Clayton et al 2007: https://www.nature.com/articles/nature05574
Piedrafita et al 2020

In 2007 a *lineage tracing* experiment was performed. In this experiment a mouse was engineered so that its cells included an inducible, inheritable label, to allow single cells and their descendents (known as clones) to be tracked and examined. This is performed in mature mice, so the tissue is at homeostasis with a constant number of stem cells over time.

![image.png](Images/Protocol.png)

*Lineage tracing experiment. Mice are genetically engineered with an inducible, inheritable label. A small number of cells in the mouse are induced, and the number of descendents bearing that label are examined later in life. Figure taken from Piedrafita et al, 2020*

This experiment had two distinct stages. First a small subset of cells were induced to express the label. This needed to be a very small fraction to minimise the chances that neighbouring cells would be induced. Second at a later timepoint, the tissue was examined and the cells with a label studied. As the level of induction is intentionally kept low, labelled cells that are adjacent can be assumed to have all grown from a single cell at induction and so the number of cells can be counted to tell you how one cell grew. Each measurement can only be performed once, so you cannot track a single clone at multiple timepoints.

![image.png](Images/Cartoon.png)

![image.png](Images/Experiment.png)

*Tracking clones with lineage tracing. An inheritable label is induced in single cells at the start of the experiment, and in subsequent examinations the number of cells are counted. The cartoon indicates how this is expected to work, whilst microscopy images with labels for all nuclei (blue) and the inducible label (red) show what is studied.*

Different types of data can be measured from these clone sizes at different times from induction;

1) How many cells are present in an area
2) How many clones are present in an area
3) How the clone size changes over time

These tell you;

1) Do the cells represent the wider tissue homeostasis (is the total density of cells constant over time?)
2) How many clones are lost over time
3) How do the surviving clones grow

What was found was that the total cell numbers were constant, but that many clones failed to persist. The loss of clones was compensated for by a linear growth in the surviving clones that can be explained by a simple mathematical model where there was only one population of dividing cells. These features (homeostasis, survival, clone growth) were subsequently found in other squamous epithelial tissues. This observation retrospectively explained observations that a small number of clones persisted throughout the life of the tissue, as clones are constantly lost (at a slowing rate) over time. These observations led to this mathematical model- known as the single progenitor model, as there was only a single type of dividing cell- to replace other models of stem cell homeostasis. It was confirmed in other tissues and later in new experiments that allowed clones to be tracked over time through "live imaging" microscopy experiments.

![Screenshot 2026-07-10 at 16.46.44.png](images/Muller.png)

*Muller plot showing clonal change over time. At the start of the simulation there are 10000 cells draw as a row, each with a different colour. At each timepoint the rows are redrawn based on which clones have survived or grown. Over time thin, tightly packed lines are replaced with fatter lines as some clones become extinct and other grow.*

These features can be understood as a result of genetic drift arising from stochastic effects of competition in the tissue. Every clone has an equal chance of survival at the start of the experiment, but by random chance some grow and some are replaced. Cells in clones that are lost are made up for by expansion of others clones, whilst the total population of cells that are labelled remains constant.

## Modelling genetic drift and non-neutral expansion

To explore the implications of genetic drift we can build a *computational model* of division and death in dividing cell populations. This expresses the processes that we know take place in a competitive environment- cells dividing and replacing their neighbours, cells being replaced by their neighbours- and turns them into a computer program that can be run and studied. 

Computational models have several advantages over experiments. By taking our interpretations and placing them into a simulation we can test and confirm that they work as expected, reproducing experimental observations. We can use this not only to show if our view is right or wrong, but also to explore alternative hypotheses and test uncertainties. Finally, we can use our models to make *predictions* for properties we have not measured experimentally. These can be used to design new experiments, or to relate our models and previous observations to other data sets and observations not originally considered. In terms of clinical implications, this might include exploring how a drug could work, or whether a genes mutation status can be interpreted as a biomarker and used to assist patient diagnosis or prognosis.

There are a diverse set of different computational models in the literature, and different modelling approaches. All models can however be understood through two key features

1) State- How we describe the status of different elements of the system (e.g. what is the concentration of different biomolecules?)
2) Update- How models state changes in each step of a simulation (e.g. how does the concentration of a biomolecule change over a time period?)

In the models of evolution we are studying here, the state of the system is the number of different types of cell in a population of fixed size. An update describes how cells die and divide, and depends on the type of model chosen.

Here we are introducing two such models, and then describing how we can make small changes to better represent epithelial tissues that we will study in other notebooks in this practical. To help us understand how the models update, before we run them in the computer, in each case we will simulate the systems using a pen/paper (or a spreadsheet) and a die (or google "d6 die" to get access to a simulator).

### Wright-Fisher models

In the Wright-Fisher models all cells in a population are replaced in a single update. If all cells have an equal fitness, then the next generation is created from cells randomly drawn from the previous generation. If we have a simple system of four cells, with clonal identity A B C and D, an example update could look as below.

| Old      | New |
| -------- | ------- |
| A  | A    |
| B | A     |
| C    | B    |
| D | A|

For a large enough population made up of equal amounts of A B C and D type cells, we would expect their overall levels to remain stable. On a per-cell basis however we would see individual clones expand and be replaced as seen in this example. Drift can therefore be seen to have a higher impact in smaller systems where extinction (complete loss of clones) or takeover (replacement of the whole population by a single clone) occurs more quickly.

In this model we have several intrinsic assumptions. The cell population is constant, every cell is replaced simulatanously, and the new generation can be made up of multiple descendents of single cells in the previous generation. All of these assumptions can be interrogated; some may be adequate simplifications but others may change the dynamics (as we see when we compare results to the Moran process described below). In our model, all cells also have equivalent fitness. This need not be the case as we describe later.

Here we will perform simulations using a die and count how many generations it takes for a subpopulation to disappear. We will assume you are using a six sided dice, but this can work with any dice as long as your system has as many cells as sides on the dice. If you use a spreadsheet, you can use a random number generator to have a system as large as you want! Remember though that a larger system will take longer to reach *fixation*- the end state where one cell type has taken over.

First, create a table with six rows. These represent cells. Label half of them with "A" and half of them with "B". Create a header above the rows to represent the time starting at zero

Second, roll the dice once for each row. The number you roll tells you which cell is surviving into the next population. So if you roll a one, then add the label from the first row to the second column. Each new cell is drawn from the previous population.

An example is given below. At time zero we have six cells, three As and three Bs. You roll the dice six times getting a 3, 2, 6, 3, 4, 1, giving you this table.

| 0      | 1 |
| -------- | ------- |
| A  | A    |
| A | A     |
| A | B     |
| B | A|
| B | B|
| B | A|

Repeat this process until all of one cell type has been replaced- that is, all cells are A or B. Do this five times and count how many updates it takes.

A muller plot is a graph with time on the x axis and clonal composition (i.e. the fraction of cells that are A, B, C etc) drawn as an aligned, stacked bar chart, allowing you to visualise how individual clones grow or are lost over time. Rather than having the normal flat top of a bar chart, the heights of the peaks are joined so you can observe the expansion and loss clearly. 

This example simulation illustrates the muller plot. In the first step A and B maintain a single cell in the next generation, E and F are lost, and C and D expand to two cells. We colour A red, B green, C white, D blue, E lime, and F purple. 

| 0      | 1 |
| -------- | ------- |
| A  | B   |
| B | A     |
| C | D     |
| D | C|
| E | D|
| F | C|

At t=0 in the Muller plot, all bars are one cell thick. At t=1, lime and purple go to zero, red and green remain constant, and white and blue expand. Note the order of clones doesn't change between time points, in order to visualise the change over time.

### Moran-process models

A Moran process is a variation of the Wright-Fisher model. The states of the model are the same- a fixed population of cells. However, now an update picks a single cell at random to be replaced, and a single cell to replace that cell. This allows different orders of replacements to occur, and as a result increases the drift of the system. 

This model has similar but distinct intrinsic assumptions compared to the Wright-Fisher model. The cell population is constant, but in contrast to Wright-Fisher only cell is replaced at a time. If we consider a generation to be the number of divisions required to replace the same number of cells in the total population, then one cell can contribute multiple descendants to the next generation, or none. How this impacts the model dynamics we can see below.

To explore this we will again simulate with a die. 

First, as before, create a table with six rows. These represent cells. Label half of them with "A" and half of them with "B". Create a header above the rows to represent the time starting at zero

Second, roll the die once. The number you roll tells you which cell is being replaced. Roll the die a second time, and this cell is the one replacing the first. They can be the same cell! So if you roll a one, then a two, then in the second column copy the first column, exchanging the label in the first row rolled with the label in the second.

An example is given below. At time zero we have six cells, three As and three Bs. You roll the dice twelve times getting a (3, 2), (6, 3), (4, 1), (1, 1), (2, 5) and (3, 4) giving you this table. Changes are highlighted.

| 0 | 1 | 2 | 3 | 4 | 5 | 6 |
| - | - | - | - | - | - | - |
| A | A | A | A |*A*| A | A |
| A | A | A | A | A |*B*| B |
| A |*A*| A | A | A | A |*A*|
| B | B | B |*A*| A | A | A |
| B | B | B | B | B | B | B |
| B | B |*A*| A | A | A | A |

These six steps are the equivalent of one step in the Wright-Fisher model, as it takes six steps to perform six replacements here whilst in Wright Fisher all six cells are replaced in a single step. Run this simulation five times and count how long it takes for one subpopulation to be replaced. To compare it with Wright Fisher steps, divide the time by six.

Whilst this is random, because of the increased alternative orderings, we expect on average Moran processes to drift at twice the speed of Wright-Fisher processes. As a result, we expect it to take half the time to reach fixation.

Simulate it for the equivalent number of steps used for clonal simulations in the Wright Fisher model above (i.e. the number of steps taken, multiplied by 6 to make Wright Fisher generations and Moran steps equivalent) and then draw a Muller plot of the clonal survival and expansion over time. ### Which model is best?

The Wright-Fisher model is much more simple, replacing all cells in a single step and using only six dice rolls to do it. So why not use this all the time? The choice of model depends on the biological system you're modelling, and its genetic drift specifically. If it is plausible that all cells divide simultaneously then the Wright Fisher may be more appropriate. This might include systems that divide according to tightly controlled circadian rhythms e.g. if all cells divide at a specific time in the morning. 

However many systems are not tightly coupled, with cells independently dividing. This makes the Moran process more appropriate, particularly considering that the drift may be unnaturally slower in Wright Fisher models as a result. However, as detailed below, other common model variations may make drift more or less important and make using the Wright Fisher approach an adequate approximation.

## Important variations

### Non neutrality

As we age, our cells gain mutations. This can be a result of an external mutagen such as UV light on the skin for example. This can also arise from intrinsic processes too, as the normal process of cell division can cause mutation. Some of these mutations have no effect on the cell, whilst others can create selective advantage or disadvantages. Selective advantages can allow clones to colonise and take over the tissue over time, and when we study the tissues of older people we find that they are a mosaic of different mutant clones that have expanded as a result of fitness advantages. 

We can introduce this in our models by weighting the selection of cells in replacement steps according to fitness. In the Wright Fisher model, the selection of cells from the previous generation to the next is random but with a higher probability of selection given to those with greater fitness. In the Moran process, the fitness typically weights the selection of the dividing cell, whilst the cell being replaced is selected wholly at random.

Takeover of the tissue by mutant clones is less sensitive to drift than neutral competition is, due to the exponential expansion driven by fitness advantage. When considering takeover in non-neutral systems the approximation taken by choosing a Wright Fisher model is more robust.

#### Simulating a non-neutral system with a die

Following the Wright Fisher model example above we are going to give "A" cells a small fitness advantage. We will follow the die scheme above except when a B cell is selected to join the next generation, we are going to roll a second die. If we roll a 1, then the B isn't successful and you need to reroll for replacement. This way we introduce a fitness disadvantage for B cells. 

The full scheme is below for a six sided dice.

First, create a table with six rows. These represent cells. Label half of them with "A" and half of them with "B". Create a header above the rows to represent the time starting at zero. 

Second, roll the dice once for each row. The number you roll tells you which cell may survive into the next population. So if you roll a one, then add the label from the first row to a second column. This column should not be headed with a time as we need to confirm each row. 

Third, for each cell in the preliminary column we need to confirm that they enter the next population. For each A cell, copy it to the next population. For every B cell roll again. This roll represents the selective disadvantage. If you roll a 1, the B cell has failed to replace the cell and you need to roll to pick a different row for the replacement.

An example is given below. At time zero we have six cells, three As and three Bs. You roll the dice six times getting a 3, 2, 6, 3, 4, 1, giving you this table. We record the preliminary results in a column called x. Rows 3 and 5 are B cells and we need to confirm that they enter the next population. We roll a 4 for row 3, so the B proceeds to the next population. For row 5 we row a 1, so the B cell fails to proceed and we reroll for the row, getting a 1. At time 0, row 1 is an A cell so in the next population becomes an A cell. If we had rolled a 3-6 (all B cells at time 0) this would have become a B cell without a reroll).

| 0      | x | 1 |
| -------- | ------- | ------ |
| A | A    | A |
| A | A     | A  |
| A | B     | B |
| B | A| A  |
| B | B| A |
| B | A| A |

Repeat this process until all of one cell type has been replaced- that is, all cells are A or B. Do this five times and count how many updates it takes and how often A takes over the whole population. 

Repeat this changing the protocol so that a 1 or 2 is sufficient to reject a B replacement. How often does A take over, and how long does it take?

### Space



*A hexagonal array of cells. The light blue cell is adjacent to the dark blue cells and so can only replace cells in the blue neighbourhood*

In many tissues in the body, like the skin, cells only compete with immediate neighbours within the structure of the tissue. In all models described so far, each replacement cell could be any cell in the previous population. If cells reside in a tissue like the skin or the oesophagus, they exist in a hexagonal array and can only replace their immediate neighbours. This changes how clones grow as they are more likely to replace cells in the same clone that are immediately adjacent than they are in a non-spatial system. As a result, both neutral and non-neutral clones grow more slowly than they do in the non-spatial model.

In the colon, stem cells form their own small compartments known as colonic crypts. In each crypt a ring of six stem cells divides to support the structure. In this environment each cell can only replace a neighbouring cell. We will simulate this with a die by specifying that a roll of 1-3 indicates the cell in the row above replaces the cell, and a roll of 4-6 indicates the cell in the row below replaces the cell. Row one is considered to be below row six to form the ring.

First, create a table with six rows. These represent cells. Label half of them with "A" and half of them with "B". Create a header above the rows to represent the time starting at zero

Second, roll the dice once for each row. The number you roll tells you which cell from the previous population is surviving into the next population. It must be an adjacent cell, and so 1-3 indicates the row above and 4-6 the row below. So if you are looking at row 2, and roll a one, then add the label from the first row to the second column at row 2.

An example is given below. At time zero we have six cells, three As and three Bs. You roll the dice six times getting a 3, 2, 6, 3, 4, 1. This can be read as the replacing cell for each row coming from above (row 6 is considered as above row 1), above, below, above, below, above

| -------- | ------- |
| A  | B   |
| A | A     |
| A | B     |
| B | A|
| B | B|
| B | B|
